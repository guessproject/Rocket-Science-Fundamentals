# Dynamic Systems and State Transition Matrix

A dynamic system is a mathematical model of an object or process whose variables evolve with time according to a set of deterministic or stochastic laws. In engineering, dynamic systems arise whenever the current condition of a physical or computational process influences its future evolution. Examples include spacecraft trajectories, flexible structures, electrical networks, thermal systems, guidance laws, chemical processes, and estimation algorithms.

## The Concept of State

The behavior of a dynamic system is described through its **state**. The state vector can be defined as set of variables that contains all necessary information to describe the system at any given time $t$.

$$  \mathbf{x}(t)\in\mathbb{R}^{n}, \mathbf{x}(t) = \begin{bmatrix} x_1(t) & x_2(t) & ... & x_n(t) \end{bmatrix}^{T} $$

, where each $x_n(t)$ is called **state variable**.


For example, the position of a spacecraft alone is not sufficient to predict its future trajectory. Its velocity must also be known. A typical Cartesian orbital state is therefore

$$\mathbf{x}(t)= \begin{bmatrix} \mathbf{r}(t)\\ \mathbf{v}(t) \end{bmatrix} = \begin{bmatrix} x & y & z & v_x & v_y & v_z \end{bmatrix}^{T}$$

To describe how a state evolves over time, the behaviour of the dynamic system has to be described by one or more **differential equations**, depending on the number of the state variables. A differential equation is a mathematical statement that relates an unknown function to one or more of its derivatives, describing how a quantity changes in space or time:

$$ \dot{\mathbf{x}}(t) = \mathbf{f}\bigl(t,\mathbf{x}(t\bigr) $$

## Linear and Non-Linear Systems

A system is defined **linear** when the dynamics can be expressed as a set of linear differential equations. The system's outputs are proportional to the inputs, and the equations governing the motion can be written as a linear combination of the state variables. If those requirements are not met, the system is **non-linear**.

### Linear Systems and State Transition Matrix

#### A Linear System example

An example of a linear system is the harmonic oscillator with linear damping:

<center><img src="Images/spring.png" alt="Drawing" style="width: 450px;"/><center>

$$m\frac{d^2 x}{d^2 t} = -kx - \nu\frac{dx}{dt} $$

<center><img src="Images/spring_plot.png" alt="Drawing" style="width: 850px;"/><center>

The equation above describes how $x$ (displacement of the mass $m$ respect to its position at rest) and $v=\dot x$ (its velocity) vary with $t$, depending on parameter's value $k$ and $\nu$. 

The system is modeled by a second order differential equation, so we can write it as a system of two, first order differential equations. For the example above we define 

- $x_1=x$
- $x_2 = \dfrac{dx_1}{dt}$

as the the **state vector** $\mathbf{x}$ of the dynamic system:

$$\textbf{x}=\begin{bmatrix}
x_1\\ 
x_2\\ 
\end{bmatrix}$$

so, we can get the following system in vectorial form:

$$ \frac{d}{dt}\begin{bmatrix}
x_1\\
x_2
\end{bmatrix}=\begin{bmatrix}
0 &  1\\
-\dfrac{k}{m} &  -\dfrac{\nu}{m}\\
\end{bmatrix}\begin{bmatrix}
x_1 \\ x_2
\end{bmatrix}$$

or, in compact notation,

$$ \dot{\textbf{x}}(t)=\textbf{A}\textbf{x}(t) $$

In the above notation, $ \mathbf{A} ∈ \mathbb{R}^{n*n} $ is the **state (system) matrix**. It relates the time evolution model (defined by the differential equations) with the state by mean of partial derivatives; in this case it is constant because the parameters $k/m$ and $\nu/m$ do not change over time. This is an example of **time invariant** dynamic system, when the coefficients of the dependent variable and its derivatives do not change with time. If this is not the case, the system is **time variant**.

Because the system is linear, the state matrix $\bold A$ can be obtained as the **Jacobian** of the dynamics equation:

$$\mathbf A = \frac{\partial\mathbf f}{\partial\mathbf x} = \begin{bmatrix} \dfrac{\partial \dot{x}}{\partial x} & \dfrac{\partial \dot{x}}{\partial v} \\[3mm] \dfrac{\partial \dot{v}}{\partial x} & \dfrac{\partial \dot{v}}{\partial v} \end{bmatrix} = \begin{bmatrix} 0&1\\[2mm] -\dfrac{k}{m}&-\dfrac{\nu}{m} \end{bmatrix}$$


The Jacobian is the matrix of  partial derivatives of the state equations and captures the local sensitivity of the state change to the current state. 

#### General Case
In general, it is possible to write:

$$ \boxed{\dot{\textbf{x}}(t)=\textbf{A}(t)\textbf{x}(t) }$$

The **state transition matrix**

$$\boldsymbol{\Phi}(t,t_0)$$

is defined as the matrix that maps the state at the initial time $t_0$ into the state at time $t$:

$$\boxed{ \mathbf{x}(t) = \boldsymbol{\Phi}(t,t_0)\mathbf{x}(t_0) }$$

$\boldsymbol{\Phi}(t,t_0)$ is the solution of the matrix diffential equation (obtained from the two boxed equations above):

$$ \boxed{ \frac{\partial \boldsymbol{\Phi}(t,t_0)}{\partial t} = \mathbf{A}(t)\boldsymbol{\Phi}(t,t_0) }$$

, subject to the initial condition

$$\boldsymbol{\Phi}(t_0,t_0) = \bold{I}$$

Also, the state transition matrix satisfies the following important properties:


$$\boldsymbol{\Phi}(t_2,t_0) = \boldsymbol{\Phi}(t_2,t_1) \boldsymbol{\Phi}(t_1,t_0)$$
$$\boldsymbol{\Phi}^{-1}(t,t_0) = \boldsymbol{\Phi}(t_0,t)$$

The composition property means that propagation can be divided into consecutive intervals. Notice the order of matrix multiplication: the transition from $t_0$ to $t_1$ occurs first, followed by the transition from $t_1$ to $t_2$.

Each element of the STM can be interpreted as a **state sensitivity**:

$$\Phi_{ij}(t,t_0) = \frac{\partial x_i(t)} {\partial x_j(t_0)}$$

Thus, the $j$-th column of $\boldsymbol{\Phi}$ describes how a unit change in the $j$-th initial-state component affects all state components at time $t$.

If $\mathbf{A}$ is constant, the system is linear time invariant, or LTI. The state transition matrix becomes

$$\boxed{ \boldsymbol{\Phi}(t,t_0) = e^{\mathbf{A}(t-t_0)} }$$

, where the matrix exponential is defined by

$$e^{\mathbf{A}\Delta t} = \mathbf{I} + \mathbf{A}\Delta t + \frac{\mathbf{A}^{2}\Delta t^{2}}{2!} + \frac{\mathbf{A}^{3}\Delta t^{3}}{3!} +\cdots$$

Unlike the scalar exponential, this expression accounts for coupling between the state variables.

For example, consider one-dimensional motion with constant velocity:

$$\dot{x}=v, \qquad \dot{v}=0$$

In state-space form,

$$\frac{d}{dt} \begin{bmatrix} x\\v \end{bmatrix} = \begin{bmatrix} 0&1\\ 0&0 \end{bmatrix} \begin{bmatrix} x\\v \end{bmatrix}$$

Since $\mathbf{A}^{2}=0$, the exponential series terminates:

$$\boldsymbol{\Phi}(t,t_0) = \mathbf{I}+\mathbf{A}(t-t_0) = \begin{bmatrix} 1&t-t_0\\ 0&1 \end{bmatrix}$$

In the linear case, the STM propagates the complete state exactly.

### Jacobian vs. State Transition Matrix

The Jacobian is the matrix of the partials of a vectorial function respect to the state variables, and is calculated at a specific time:

$$ \textbf{J}=\left.\begin{matrix}
\dfrac{\partial \textbf{f}}{\partial x}\end{matrix}\right|_{t^*} $$

For a non-linear system (see below), the Jacobian tells how $\bold f$ changes respect to $\bold x$ at any time, and provides a linearized version of the system around that point at that time. It is a static snapshot of the non-linear system, and changes in time.

The STM describes how the state of a linear (or linearized) system evolves from $t_0$ to $t$. For LTI systems $\boldsymbol{\Phi}(t,t_0) = e^{\bold A (t-t_0)}$, where $\bold A$ is the state matrix (often obtained as Jacobian in the linearized case). It is then the result of the dynamic's integration  over time; it answer questions like *how the state evolves after a specific time interval*, and not just *how it changes instantaneuosly*.

To summarize, for a linear system the state transition matrix is the exact operator that propagates the state, whereas for a nonlinear system it is the first-order sensitivity of the flow along a specified reference trajectory.

### Non-linear Systems

A non-linear system is a system in which the change of the output is not proportional to the change of the input. Typically, the behavior of a non-linear system is described in mathematics by a non-linear system of equations, where the equation(s) to be solved cannot be written as a linear combination of the unknown variables or functions that appear in them. As non-linear dynamical equations are difficult to solve, non-linear systems are commonly approximated by linear equations (**linearization**). 

Linearization of a 1-D function means using the tangent line of a function at a point as an approximation to the function in the vicinity of the point; similarly, linearization of a 2-D function means using the tangent plane:

#### A Non-linear System Example

An example of non-linear problem is the dynamics of a frictionless pendulum under the influence of gravity:
<center><img src="Images/pendulum.png" alt="Drawing" style="width: 200px;"/><center>

 The differential equation which governs its motion is


$$ \frac{d^2 \theta}{d^2 t} = -\frac{g}{l} \sin \theta $$

<center><img src="Images/pendulum_plot.png" alt="Drawing" style="width: 850px;"/><center>

We define the state vector as:

$$\textbf{x}=\begin{bmatrix}
\theta\\ 
\dot \theta\\ 
\end{bmatrix}=\begin{bmatrix}
\theta\\ 
\omega\\ 
\end{bmatrix}$$

Decomposing the differential equation we have:

$$ \begin{matrix}
\dfrac{d}{dt} \theta = \omega \\
\dfrac{d}{dt} \omega = -\dfrac{g}{l} \sin \theta
\end{matrix} $$

Since $\sin \theta$ is not a linear operator, it is not possible to relate the state vector derivative to the state itself in the form $ \dot{\textbf{x}}(t) = \textbf{A}\textbf{x}(t)$ (the state is inside the $\bold A$ matrix):

$$\mathbf{A}(t) = \frac{\partial\mathbf{f}}{\partial\mathbf{x}} \bigg|_{\bar{\mathbf{x}}(t)} = \begin{bmatrix} 0&1\\ -\dfrac{g}{l}\cos{\theta}&0 \end{bmatrix}$$



While some time we find the solution by analytically integrating the equation of motion, the most common way to solve this issue is to **linearize** the equations of motion in the surrounding of a particular state (*initial conditions*).

In the above case, we find that $\sin \theta \approx \theta$ for $\theta \approx 0$, so for small $\theta$ we have

$$ \frac{d^2 \theta}{d^2 t} = -\dfrac{g}{l} \theta $$

In this case the matrix $\bold A$ becomes constant:

$$\mathbf{A} = \begin{bmatrix} 0&1\\ -\dfrac{g}{l}&0 \end{bmatrix}$$

...and the STM is obtained from

$$\boldsymbol{\Phi}(t,t_0)=e^{\bold A (t-t_0)}  = \begin{bmatrix} \cos\!\left(\sqrt{\dfrac{g}{l}}\,(t-t_0)\right) & \dfrac{1}{\sqrt{g/l}}\, \sin\!\left(\sqrt{\dfrac{g}{l}}\,(t-t_0)\right) \\[10pt] -\sqrt{\dfrac{g}{l}}\, \sin\!\left(\sqrt{\dfrac{g}{l}}\,(t-t_0)\right) & \cos\!\left(\sqrt{\dfrac{g}{l}}\,(t-t_0)\right) \end{bmatrix}$$


This is the analytical representation of the STM. However, this local linear system should not be confused with the original nonlinear pendulum dynamics. The STM describes the local evolution of perturbations, not the entire nonlinear trajectory.

#### General Case

In general, for non-linear systems there is no matrix $\boldsymbol{\Phi}$, independent of the initial state, such that

$$\mathbf{x}(t) = \boldsymbol{\Phi}(t,t_0)\mathbf{x}_0$$

for every $\mathbf{x}_0$.


Let

$$\bar{\mathbf{x}}(t)$$

be a nominal solution satisfying

$$\dot{\bar{\mathbf{x}}}(t) = \mathbf{f}\bigl(t,\bar{\mathbf{x}}(t)\bigr)$$

Consider a nearby trajectory

$$\mathbf{x}(t) = \bar{\mathbf{x}}(t)+\delta\mathbf{x}(t)$$

Substituting into the nonlinear dynamics gives

$$\dot{\bar{\mathbf{x}}} + \delta\dot{\mathbf{x}} = \mathbf{f} \left( t, \bar{\mathbf{x}}+\delta\mathbf{x}\right)$$

After a first-order Taylor expansion about the nominal trajectory and nominal equation subtraction, the perturbation dynamics become

$$\boxed{ \delta\dot{\mathbf{x}}(t) = \mathbf{A}(t)\delta\mathbf{x}(t) }$$

where

$$\mathbf{A}(t) = \left. \frac{\partial\mathbf{f}} {\partial\mathbf{x}} \right|_{\bar{\mathbf{x}}(t)}$$

The first-order evolution is

$$\boxed{ \delta\mathbf{x}(t) \approx \boldsymbol{\Phi}(t,t_0) \delta\mathbf{x}(t_0) }$$

, where

$$\boxed{ \dot{\boldsymbol{\Phi}}(t,t_0) = \mathbf{A}(t)\boldsymbol{\Phi}(t,t_0), \qquad \boldsymbol{\Phi}(t_0,t_0)=\mathbf{I} }$$


<center><img src="Images/perturbation.png" alt="Drawing" style="width: 650px;"/><center>


This is the system of **first-order variational equations**. The state transition matrix can be used for any non-linear dynamical system and is used to approximate the dynamics of a system over short period of times. Its components are the partial derivatives of the state at time $t$ with respect to the state at time $t_0$

$$ \mathbf{\Phi}(t,t_0)=\dfrac{\partial  \mathbf{x}(t)}{\partial \mathbf{x}(t_0)}, \quad \mathbf{\Phi} \in \mathbb{R}^{n*n}$$


**A concrete interpretation example**

Suppose the propagated STM is

$$\boldsymbol{\Phi}(t_f,t_0) = \begin{bmatrix} 1.20 & 0.10\\ 0.35 & 0.90 \end{bmatrix}$$

For an initial perturbation

$$\delta\mathbf{x}_0 = \begin{bmatrix} 0.01\\ -0.02 \end{bmatrix}$$

the final perturbation is

$$\delta\mathbf{x}_f = \boldsymbol{\Phi}\delta\mathbf{x}_0 = \begin{bmatrix} 1.20 & 0.10\\ 0.35 & 0.90 \end{bmatrix} \begin{bmatrix} 0.01\\ -0.02 \end{bmatrix} = \begin{bmatrix} 0.010\\ -0.0145 \end{bmatrix}$$

This means:

- the final first-state error is produced by contributions from both initial errors;
- the final second-state error is also coupled to both initial variables;
- the off-diagonal elements quantify cross-coupling;
- the STM maps the full initial perturbation vector, not each error independently.

### Linear vs. Non-linear

**Linear Systems**

For a generic linear system

$$\dot{\mathbf{x}}=\mathbf{A}(t)\mathbf{x}$$

the relation

$$\mathbf{x}(t) = \boldsymbol{\Phi}(t,t_0)\mathbf{x}(t_0)$$

is exact for every admissible initial state.

The same STM applies throughout the state space because $\mathbf{A}$ does not depend on $\mathbf{x}$.

**Non-linear Systems**

For a generic non-linear system

$$\dot{\mathbf{x}}=\mathbf{f}(t,\mathbf{x})$$

the nominal state must be propagated through the nonlinear equations:

$$\mathbf{x}(t) = \boldsymbol{\varphi} (t,t_0,\mathbf{x}_0)$$

while

$$\boldsymbol{\Phi}(t,t_0;\mathbf{x}_0) = \frac{\partial\boldsymbol{\varphi}} {\partial\mathbf{x}_0}$$

describes the local first-order sensitivity of that nonlinear flow $\boldsymbol{\varphi}$ to the initial conditions. So, the STM propagates only a first-order perturbation:

$$\delta\mathbf{x}(t) = \boldsymbol{\Phi} (t,t_0;\bar{\mathbf{x}}_0) \delta\mathbf{x}_0 + \mathcal{O} \left( \|\delta\mathbf{x}_0\|^{2} \right)$$


Consequently, nonlinear systems do not possess one universal STM valid everywhere in state space. Instead, each nominal trajectory generates its own time-varying linear perturbation model and therefore its own STM. This is precisely what makes the STM useful: it translates nonlinear dynamics, locally and systematically, into a linear form that engineers can use for sensitivity analysis, estimation, uncertainty propagation, trajectory correction, optimization, and stability analysis.



Therefore:

- the STM depends on the nominal trajectory;
- it allows the propagation of small initial-state errors;
- the mapping is local rather than globally exact;
- its accuracy decreases as the initial perturbation grows;
- its accuracy may also degrade over long propagation intervals;
- relinearization or an iterative correction process may be required.



### Numerical propagation of the STM

For an $n$-state non-linear model, numerical propagation typically integrates:

- $n$ equations for the nominal state;
- $n^2$ equations for the STM.

The augmented state is

$$\mathbf{z}(t) = \begin{bmatrix} \mathbf{x}(t)\\ \operatorname{vec}\boldsymbol{\Phi}(t,t_0) \end{bmatrix}$$

With dynamics

$$\dot{\mathbf{x}} = \mathbf{f}(t,\mathbf{x}), \dot{\boldsymbol{\Phi}} = \mathbf{A}(t)\boldsymbol{\Phi}$$

and initial conditions

$$\mathbf{x}(t_0)=\mathbf{x}_0, \qquad \boldsymbol{\Phi}(t_0,t_0)=\mathbf{I}$$

For a six-state dynamic problem, this means integrating $6+62=426+6^{2}=42$ scalar differential equations. The additional equations are not intended to improve the nominal state propagation itself. Their value is that they provide the complete first-order derivative information surrounding that nominal solution. 



## Stability Information and Monodromy Matrix

A large STM element means that some final-state component is very sensitive to some initial-state component, but individual entries alone are coordinate-dependent and should not generally be used as a complete stability criterion.

More robust indications come from:

- singular values of the STM;
- eigenvalues for appropriate fixed-point or periodic-orbit mappings;
- the condition number;
- the monodromy matrix for periodic trajectories.

For a periodic orbit with period $T$, define

$$\boxed{ \mathbf{M} = \boldsymbol{\Phi}(t_0+T,t_0) }$$

where $\mathbf{M}$ is the **monodromy matrix**.

Its eigenvalues, called characteristic multipliers, describe how perturbations evolve after one complete period:

$$\mathbf{M}\mathbf{v}_i = \lambda_i\mathbf{v}_i$$

Interpretation:

- $\lambda_i|>1$: perturbations along $\mathbf{v}_i$ grow from one period to the next;
- $\lambda_i|<1$: perturbations contract;
- $\lambda_i|=1$: neutral first-order behavior;

complex unit-modulus multipliers indicate oscillatory, non-growing first-order modes.
