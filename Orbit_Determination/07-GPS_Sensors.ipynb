# GPS

## System Description

The GPS (**Global Positioning System**) sysem is composed by three segments:

- **Space Segment**: this consists of at least 24 satellites in 6 orbital planes orbiting the Earth, each equipped with highly accurate atomic clocks. The constellation is designed to ensure that the users will have, at least, 4 simultaneous satellites in view from any point at the Earth surface at any time 
- **Control Segment**: a worldwide ground control/monitoring network monitors the health and status of the satellites. This network also uploads navigation and other data to the satellites
- **User Segment**: the GPS receiver onboard vehicles, mobile phones etc.

 <center><img src="Images/GPS_segments.png" alt="Drawing" style="width: 600px;"/><center>




Each GPS satellite broadcasts ranging codes and navigation data:
- the **navigation message** include ephemeris of the transmitting satellite and information about the time and status of the entire satellite constellation, called the almanac. Provides the means for the receiver to determine the location of the satellite at the time of signal transmission
- the **ranging signal** contain a Pseudo Random Noise (**PRN**) code that allows satellite identification and ranging estimation. The original GPS design contains two ranging codes: the Coarse/Acquisition (C/A) code which is modulated onto the L1 carrier and is freely available to the public, and the restricted Precision (P) code which is modulated onto the L1 and L2 carriers allowing for the removal of the first order effects of the ionosphere.

GPS satellites trasmit on three different frequencies:

- **L1** Frequency: 1575.42 MHz This is the primary frequency used by GPS for navigation and timing. The L1 frequency is used to transmit the C/A (Coarse/Acquisition) and P(Y) code signals. The C/A code is a civilian code that is available to all GPS users, while the P(Y) code is a military code that is encrypted and only available to authorized users
- **L2** Frequency: 1227.60 MHz This frequency is used to transmit the P(Y) code signal, which is a military code that is encrypted and only available to authorized users. The L2 frequency is also used to transmit the L2C signal, which is a civil signal that is available to all users, it’s more precise than the C/A code and L1 frequency
- **L5** Frequency: 1176.45 MHz This frequency is used to transmit the L5 signal, which is a civil signal that is available to all users. L5 signal is intended for high precision navigation and timing applications and it’s more resistant to interference and multipath

On the user side, there are two measurement types available:

<font size="4">**Pseudoranges Measurements**<font>

This is the more common method used by consumer devices like smartphones and car navigation systems. The concept of one-way **Time Of Arrival** (TOA) ranging is used to determine user position. This concept entails measuring the time it takes a radio signal transmitted by an emitter at a known location to reach a user receiver. This time interval, referred to as the signal propagation time, is then multiplied by the speed of light to obtain the emitter to receiver distance. By measuring the propagation time of the signal broadcast from multiple emitters at known locations the receiver can determine its position.

<center><img src="Images/pseudoranges.png" alt="Drawing" style="width: 500px;"/><center>  

We define a **pseudorange** $ \tilde{\rho}$ as the measure of the difference in time between signal transmission and reception. It is computed by measuring the propagation time required for a satellite-generated ranging code to transit from the satellite to the user receiver antenna:

#### $$  \tilde{\rho}=c\left ( t_2-t_1 \right )$$

<center><img src="Images/ranging_code.png" alt="Drawing" style="width: 550px;"/><center>  


Since $t_2$ and $t_1$ (which are the receiving and transmitting times respectively) come from two different clocks which are not perfectly synchronised, the pseudoranges contain errors which need to be corrected. The Figure below shows how the clocks errors impact the range estimation:

<center><img src="Images/pseudorange.png" alt="Drawing" style="width: 550px;"/><center> 


,where:

- $T_s$ = system time at which the signal left the satellite
- $T_u$ = system time at whcih the signal reached the user reciver
- $\delta T$ = offset of the satellite clock from system time
- $t_u$ = offset of the receiver clock from system time
- $T_s + \delta T$ = satellite clock reading at the time that the signal left the satellite
- $T_u + t_u$ = user receiver clock reading at the time the signal reached the user receiver
- $c$ = speed of light

The satellite clock offset from system time, $\delta t$, is composed of bias and drift contributions. The GPS ground-monitoring network determines corrections for these offset contributions and transmits the corrections to the satellites for rebroadcast to the users in the navigation message; these corrections are applied within the user receiver to synchronize the transmission of each ranging signal to system time in such a way $\delta t$ is corrected and removed from the unknowns. Assuming $\textbf{s}$ is the true satellite position and $\textbf{u}$ is the true user receiver position, the pseudorange equation can be written as 

#### $$\tilde{\rho }= \left\| \textbf{s}-\textbf{u}\right\|+c t_u$$

<font size="4">**Carrier Phase Measurements**<font>

This method is used for high-precision applications like land surveying, agriculture, and scientific research. Instead of using the code, it measures the phase of the carrier wave itself. A single wavelength of the GPS L1 carrier wave is only about 19 centimeters (7.5 inches), which is much smaller than a single chip in the PRN code. By measuring the fractional part of this wavelength, the receiver can determine the distance to the satellite with much greater precision.

The carrier phase is calculated by integrating the Doppler shift at the receiver side and applying advanced processing techniques to solve ambiguities; centimeter-level precision on position estimates can be reached by using carrier phase measurements.

### Navigation Solution

In order to determine user position in three dimensions ($x_u$, $y_u$, $z_u$) and the offset $t_u$, pseudorange measurements are made to four satellites resulting in the following system of non-linear equations:

$$\begin{matrix}
\tilde{\rho_1} =\sqrt{\left ( x_1-x_u \right )^{2}+\left ( y_1-y_u \right )^{2}+\left ( z_1-z_u \right )^{2}}+c t_u \\
 \tilde{\rho_2} =\sqrt{\left ( x_2-x_u \right )^{2}+\left ( y_2-y_u \right )^{2}+\left ( z_2-z_u \right )^{2}}+c t_u\\
 \tilde{\rho_3} =\sqrt{\left ( x_3-x_u \right )^{2}+\left ( y_3-y_u \right )^{2}+\left ( z_3-z_u \right )^{2}}+c t_u\\
\tilde{\rho_4} =\sqrt{\left ( x_4-x_u \right )^{2}+\left ( y_4-y_u \right )^{2}+\left ( z_4-z_u \right )^{2}}+c t_u\end{matrix}$$

Applying linearization to the above set of equations is possible to derive the four unknowns. The estimated three-dimensional position over time is called **navigation solution**. Its precision is related with the errors contained in the pseudoranges; the most significant error is the **atmospheric delay** due to the ionosphere and troposphere, which slow down and deviate the RF signals:

   
<center><img src="Images/GPS_delays.png" alt="Drawing" style="width: 450px;"/><center>

Since a single frequency receiver is subject to ionosphere effects, to mitigate the atmospheric delay the possible options are available:


- **Dual Frequency**: evaluates the signal delay by comparing the receiving times at two different frequencies
- **Single Differencing**: observing two satellites at the same time with the same receiver, the internal clock error (which affects both measurements equally) is removed. Same applies for the same satellite observed by two receivers at the same time
- **Double differencing**:  is essentially a single difference of two single differences; it is extremely powerful because it virtually eliminates both satellite and receiver clock errors


### Diluition Of Precision and Navigation Accuracy

**Dilution of Precision** (DOP) is a crucial concept in GPS positioning that quantifies the geometric strength of the satellite configuration relative to the receiver. It tells you how good the satellite geometry is for determining your position, and thus, how much any existing measurement errors will be "diluted" or amplified into your final position solution


<center><img src="Images/gpsUncertainty1.png" alt="Drawing" style="width: 500px;"/><center>


The size and shape of the uncertainty region change depending on the spatial arrangement of the satellites as seen by the receiver.

- Good Satellite Geometry (low DOP): When satellites are widely spaced across the sky (e.g., one overhead, one to the north, one to the east, one to the west), the intersection of their range spheres (or pseudorange circles in 2D) forms a sharp, small area. This means small errors in range measurements lead to small errors in the calculated position
- Poor Satellite Geometry (high DOP): When satellites are clustered together in a small portion of the sky (e.g., all to the north or all low on the horizon), the intersection of their range spheres is broad and shallow. In this case, even small errors in range measurements can result in much larger errors in the estimated position. The uncertainty is "diluted" or spread out, leading to less precise results

<center><img src="Images/gpsUncertainty2.png" alt="Drawing" style="width: 450px;"/><center>

**Navigation accuracy** depends on the internal measurements error and on the DOP:

<p style="text-align:center;"><b>nav_accuracy = measurements_error x DOP</b></p>


### Available Tracking Data Types

Thre are three different tracking data types potentially available from the onboard receiver when performing GPS-based OD:

<center><img src="Images/GPS_receiver.png" alt="Drawing" style="width: 500px;"/><center>





1) **Row observation data**
    Defined in the Receiver Independent Exchange Format (**RINEX**) format, the observation data file can contain pseudorange, carrier phase, Doppler and/or signal strenght information for each of the GNNS (not only GPS) satellites in line of sight. Using row data it is possible to reach an accuracy of 10-20 cm or better if carrrier phase observation are available. When usign this approach, the OD problem requires the knowledge of the GPS satellites ephemeris and the clock bias. Find here the link to the specifications of the RINEX 3.02 version for additional information: https://files.igs.org/pub/data/format/rinex302.pdf
   
2) **Navigation solution**
   As discussed above, from the raw pseudoranges is possible to derive the estimated cartesian position at each measurement time. The estimation process happens onboard with limited accuracy (about 2-3 meters); and the measurements do not require additional information about the GPS constellation

3) **On-board filter**
   The navigation solution provided above can be further processed using onboard Kalman filter to also derive velocity data. This option should be avaoided if possible since the filtering is done using low-fidelity models and the accuracy increases up to 20 meters

## OD with raw pseudo-ranges and phase-counts

Processing the raw pseudorange and phase directly from receiver provides significantly better results than the other options. Depending on the available data, the following options are available:

<center><img src="Images/GPS_row_meas.png" alt="Drawing" style="width: 700px;"/><center>



- **CA/P1/P2 Pseudorange**: Pseudo-range is processed using CA, P1 or P2 code on L1/L2. The range between GPS satellites and the receiver is the modeled measurement. Receiver clock phase error and clock frequency errors are optionally modeled and estimated
- **CA SD (Single Differenced) Pseudorange**: The modeled measurement is the difference in CA range measurement from two distinct GPS satellites received at the same time. In this case, receiver clock errors are removed from the measurement model, eliminating the need to estimate receiver clock phase and frequency
- **DF (Dual Frequency) Pseudorange**: Instead of processing P1 pseudo-range and P2 pseudo-range separately, the P1 and P2 pseudo-ranges are combined into one pseudo-range that is corrected using an ionospheric correction model. The receiver clock phase error and clock frequency error are optionally modeled and estimated
- **DF SD (Dual Frequency Single Differenced) Pseudorange**: In this case, each GPS satellite's P1 and P2 pair is combined into one (ionospherically-corrected) pseudo-range. The modeled measurement then becomes the range difference from two distinct GPS satellites corrected pseudo-ranges. Receiver clock phase and frequency are not estimated
- **LA/L1/L2 Phase**: Carrier phase-count on L1/L2 derived from CA/P code tracking. Receiver clock phase error and clock frequency error are optionally modeled and estimated
- **L1 SD (Single Differenced) Phase**: The modeled measurement is the phase-count difference in L1 phase-count measurement from two distinct GPS satellites received at the same time. In this case, receiver clock errors are removed from the measurement model, eliminating the need to estimate receiver clock phase and frequency
- **DF (Dual Frequency) Phase**: Instead of processing L1 phase-count separately, L1 and L2 phase -ounts are combined into a single phase-count observation corrected for ionosphere effects. The receiver clock phase error and clock frequency error are optionally modeled and estimated
- **DF SD (Dual Frequency Single Differenced) Phase**: In this case, each GPS satellite's L1 and L2 pair is combined into one (ionospherically-corrected) phase-count. The modeled measurement then becomes the phase count difference from two distinct GPS satellites corrected observations. Receiver clock phase and frequency are not estimated

### RINEX Files

The Receiver Independent Exchange Format (**RINEX**) has been designed to facilitate the exchange of GNSS data independently from the receiver's type and characteristics. RINEX files  are downloaded from the onboard receiver to allow a more accurate navigation solution process than the one performed onboard). There are three different types of RINEX files:

1) Observation data file - contains information on the pseudorage/phase count at observation time
2) Navigation message file - contains information on the GNSS constellation ephemeris and clock correction
3) Metereological data file

The RINEX observation data file has the following structure:

, where:

- 1 = GNSS constellation info (e.g., G = GPS, E = Galileo)
- 2 = Number of observations for that constellation
- 3 = Observation codes for that constellation (e.g., C1C = C/A code on L1 frequency)
- 4 = Onboard receiver time tag and observations taken at that time relevant to specific GNSS spacecrafts

### Advantages and Disadvantages
**Advantages**:
- Autonomy and Real-Time Capability: this is the most significant advantage. The satellite can determine its own orbit without relying on ground-based tracking stations, which frees up ground control resources and allows for immediate, real-time navigation. This is critical for missions requiring rapid, autonomous maneuvers
- High Accuracy: onboard GPS, especially when using dual-frequency receivers and carrier phase measurements, can achieve very high precision. Post-processed GPS data can lead to centimeter-level orbit determination for LEO satellites, rivaling or even surpassing traditional ground-based tracking networks
- Reduced Ground Station Reliance: it minimizes the need for a global network of ground stations, which reduces operational costs and complexity for mission control
- Continuous Coverage: unlike ground-based radar or optical sensors, which are limited by line-of-sight and weather, an onboard GPS receiver can continuously track multiple GPS satellites, providing a steady stream of data
- Improved Science Return: For Earth observation and remote sensing missions, highly accurate, real-time onboard orbit data is essential for precisely geo-referencing images and other measurements.

**Disadvantages**:
- LEO Orbit Constraint: onboard GPS is most effective for satellites in LEO (up to around 2,000 km altitude). For satellites in MEO or GEO, it is much more difficult to get a strong signal from the GPS constellation below, and the geometry is often not as favorable
- Signal Environment: the space environment presents unique challenges. The satellite is often "above" the GPS constellation, and the signals are not designed for this high-altitude usage. Signal strength can be weak, and antennas need to be carefully designed
- Radiation Hardening: the electronic components of the GPS receiver must be specially designed and shielded to be "radiation-hardened" to survive in the harsh space environment
- Onboard Processing Power: to run advanced filters like a Kalman filter, the satellite's onboard computer needs to have sufficient processing power
- Power Consumption: GPS receivers consume power, and on missions with strict power budgets, they may be turned on only intermittently, leading to gaps in the navigation solution
