# Measurement Model


## Sensor Classification by Observable

Although expressed through a common measurement equation $h$, different sensors types differ significantly in the observables they provide, their accuracy characteristics, coverage, operational constraints, and contribution to state observability. Sensor types are best understood in terms of the observables they provide rather than the physical technology used to collect them:

### Range Measurements

Range is one of the most intuitive observables; it is defined as 

$$\rho = ||\mathbf r_{sat}-\mathbf r_{station}||$$

, where:

- $\mathbf r_{sat}$ is spacecraft position
- $\mathbf r_{station}$ is tracking-station position


### Doppler (or Range-Rate) Measurements

The Doppler observable is

$$\dot{\rho} = \frac{(\mathbf r \cdot \mathbf v)} {||\mathbf r||}$$

It measures the relative velocity projected along the line of sight. Unlike range, doppler contains significant velocity information along theradial direction.

Because orbit propagation directly couples position and velocity, Doppler measurements often produce substantial improvements in trajectory accuracy.

### Angular Measurements

Optical sensors measure line-of-sight direction through quantities such as:

- Right Ascension
- Declination
- Azimuth
- Elevation

These observables define a direction vector:

$$\hat{\mathbf l} = \frac{\mathbf r}{||\mathbf r||}$$

The information content is fundamentally different from range, since it provides direction instead of distance. Angular measurements constrain the spacecraft to a line extending from the observer. Without additional measurements, the satellite could lie anywhere along that line: this phenomenon is commonly known as the range ambiguity of optical observations.

## Non-linearity of the Measurement Model

The relationship between the observables and the (unknown) state is often non-linear. Consider the simple case of a satellite being tracked by a ground station located at the center of a 2-dimensional reference system:

<center><img src="Images/radar.png" alt="Drawing" style="width: 420px;"/><center>


Only considering the satellite's position $(x, y$)$, the range equation is

$$ \rho(t) = \sqrt{x(t)^2 + y(t)^2}$$

...which is clearly non-linear. 
 The state to measurements matrix $\bold H$ is

 $$ \bold H =\begin{bmatrix}
\dfrac{\partial \rho}{\partial x} & 
\dfrac{\partial \rho}{\partial y} 
\end{bmatrix} = \begin{bmatrix}
\dfrac{x}{\sqrt{x^2+y^2}}& 
\dfrac{y}{\sqrt{x^2+y^2}}\end{bmatrix}$$

$\bold H$ represents the linearized model of the state to measurements relationship. Applying a Taylor's series expansion and evaluating $\bold H$ at the current state estimate we get the local dependency of the state variation respect to the measurements variation.

If we don't consider the measurement errors, the state to measurements relationship becomes

$$\rho = \bold H \bold x \qquad \bold x = \bold H^t \rho$$



### Information and State Observability

A distinction must be made between **information content** and **observability**. Information content describes how much a measurement contributes to estimating the state, while observability describes whether the available measurements are sufficient to uniquely estimate the state.

A measurement may contain information without making the entire system observable. For example, a single optical angle measurement contains information but it does not uniquely determine a six-dimensional orbital state. As additional observations are collected over time, orbital dynamics couple different state components and increase observability.

Here we introduce the **Fisher information matrix** $\mathbf F$, which is defined as a matrix that contains information about the expected amount of information that an observable random variable carries about an unknown state:

$$\mathbf F = \mathbf H^T \mathbf R^{-1} \mathbf H$$

, where:
- $\mathbf H$ is the measurement sensitivity matrix
- $\mathbf R$ is the measurement covariance matrix

The matrix reveals two key facts:
1) Sensitivity Increases Information: larger Jacobian terms increase information (measurements that strongly react to state changes are more valuable)
2) Noise Reduces Information: larger measurement uncertainty decreases information.

Consequently, the most informative observable is the one that combines strong state sensitivity and low measurement noise.
