# Optical and Radar Sensors

## Optical Sensors
Optical sensors work by capturing light using sensitive devices (such CCD or CMOS sensors) to determine the target's angular position. The images not only contain the target satellite but also a background of stars. By comparing the star patterns in the image to a known star catalog, the sensor can precisely determine its own pointing direction and orientation; this provides a precise measurement of the satellite's angular values at a specific, time-stamped moment. For visibility, the satellite must either be illuminated by the Sun (passive tracking) or by a light emitter (laser) located at the ground station (active tracking).
Once the image is taken, image processing is needed to calculate the satellite's angular data.

### Reference Frame
Optical measurements are related to angular data; they usually refer to a set of two angles (right ascension RA and declination Dec) measured respect to an inertial reference system centered at the ground station location.  The default reference frame for RA and Dec value is MEME J2000, referring to the mean equator and mean equinox of J2000 reference frame. Other reference systems can be chosen, including TEMEs (true equator and mean equinox) at different epochs.

### Illumination Conditions
To be visible for optical tracking, a satellite it must be illuminated (normally by the Sun) and, because of its small size and the difficulty to obtain a sufficient level of illumination, it must be observed against a dark sky background.

### Aberration
In astronomy, aberration is a phenomenon which produces an apparent motion of celestial objects about their true positions, dependent on the velocity of the observer. It causes objects to appear to be displaced towards the direction of motion of the observer compared to when the observer is stationary:

$\quad$
<center><img src="Images/aberration.png" alt="Drawing" style="width: 250px;"/><center>    
    
$\quad$





The change in angle is of the order of $v/c$ where $c$ is the speed of light and $v$ the velocity of the observer. There are two types of aberration:


#### Annual aberration
Annual aberration is caused by the motion of an observer on Earth as the planet revolves around the Sun. Due to orbital eccentricity, the orbital velocity of Earth (in the Sun's rest frame) varies periodically during the year as the planet traverses its elliptic orbit and consequently the aberration also varies periodically, typically causing stars to appear to move in small ellipses (whose semimajor axis is about 20.5 arcseconds).

#### Diurnal aberration
Diurnal aberration is caused by the velocity of the observer on the surface of the rotating Earth. It is therefore dependent not only on the time of the observation, but also the latitude and longitude of the observer. Its effect is much smaller than that of annual aberration, and is only 0.32 arcseconds in the case of an observer at the Equator, where the rotational velocity is greatest.

### Advantages and Disadvantages
**Advantages**:
- High Angular Precision: modern optical sensors can achieve extremely high precision in their angular measurements, often on the order of arcseconds.
- Suitable for High Orbits: optical sensors are particularly effective for tracking objects in MEO and GEO, where radar signals are weaker due to the long distances
- Passive Sensing: they are "passive" sensors that don't need to transmit their own signals. This makes them ideal for covert operations and reduces power consumption
- Cost-Effective for Deep Space: for tracking very distant objects, a large, sensitive telescope can be more cost-effective than a powerful radar system.

**Disadvantages**:
- No Range Information: the biggest drawback is that they provide "angles-only" measurements. This makes the initial orbit determination (IOD) problem more complex and often requires a longer arc of observations from multiple locations to get a good position fix
- Dependency on Illumination: most optical sensors rely on reflected sunlight to see the target. This limits their operation to twilight and nighttime hours when the satellite is still illuminated by the sun but the sensor is in darkness
- Weather and Atmospheric Effects: ground-based optical sensors are highly susceptible to cloud cover, atmospheric turbulence, and light pollution, which can severely limit their operational time and accuracy
- Obscuration: they can be blinded by bright objects like the sun or moon and can't see through clouds or dense fog

## Radar Sensors

Radar measurements are based on the principle that electromagnetic waves are reflected if they meet an electrically leading surface. If these reflected waves are received again at the place of their origin, then that means an obstacle is in the propagation direction. 

<center><img src="Images/radar1.gif" alt="Drawing" style="width: 550px;"/><center> 

$\quad$

Beside the distance, radar can estimate object's velocity thanks to the **doppler effect**. This effect describes the change in frequency or wavelength of a wave in relation to an observer who is moving relative to the source of the wave.

- Approaching Object: If the source of the waves is moving towards the observer, the waves get compressed, leading to an increase in frequency (and a decrease in wavelength)
- Receding Object: If the source of the waves is moving away from the observer, the waves get stretched out, leading to a decrease in frequency (and an increase in wavelength)

The relative velocity between satellite and observer is computed from the Doppler shift by 

$$\dot{\rho }=c\left ( 1-\frac{v_r}{v_t} \right )$$

, where $v_r$ is the received frenquency at ground station and $v_t$ is the nominal transmitter frequency.

**Angular data** (in a topocentric reference system) are given from the antenna’s boresight pointing direction at time of pulse transmission. The angular resolution is then related with the beamwidth of the transmitting antenna.
Radar observation data are given in Azimuth (or "Az") and Elevation (or "El") - Elevation is the angular distance above the horizon (0 < El < 90°), and azimuth the angular distance (0 < Az <360°), measured along the horizon, westwards from the south point S (in astronomy) or eastwards from the north point N in navigation. Az/El are local coordinates: each observer sets up his own reference frame.


### Types of Radar Systems
- **Monostatic Radar**: the transmitter and receiver are co-located. This is the most common configuration
- **Bistatic Radar**: the transmitter and receiver are separated by a significant distance. This can be more complex but offers unique advantages, such as enhanced coverage and the ability to detect objects from a different viewing geometry
- **Phased-Array Radar**: these are highly sophisticated systems with a large array of fixed antennas that can electronically steer their beam without physical movement. This allows for extremely rapid scanning of large volumes of space and simultaneous tracking of many objects, which is crucial for monitoring dense orbital regions like LEO

### Error Sources

Radar measurements are affected by several measurement errors that we’d like to summarize:
- **Atmospheric delay**: the electromagnetic signals are slowed down by the atmosphere during the propagation as they interact with electrons, ions, atoms and molecules. The measured range is then higher than in the case of absence of atmosphere. The amount of velocity reduction is function of the temperature, humidity and pressure, and two different media correction models are needed to estimate the range correction (one for the troposphere – 0 to 50 km – and one for the ionosphere – 50 to 1000 km)
- **Transponder delay**: the on-board transponder does not retransmit instantaneously the received range signal, and the delay it introduces leads to an artificial increase of the measured range. The delay is about 1000 – 3000 nsec, equivalent to a range bias of about 150-450 meters
- **Satellite motion**: since the satellite is travelling along its orbit, its position at the transponder receiving time is different than its position at the pulse transmitting time

### Two-Way vs. Three Way

'Two-way' vs. 'three-way' concept is related to the number of active participants in the communication or measurement process. A two-way radar involves a single radar transmitting and receiving its own signal, while a three-way system introduces a third party, such as a target with its own receiver or a second ground station, to process or retransmit the signal, providing additional information or enabling more accurate measurements. 

#### Two-Way Radar 

In a standard radar system, a single device (station 1) transmits radio waves and then receives the reflected signal from the spacecraft (station 2). Range is determined by the time it takes for the signal to travel to the target and return.

#### Three-Way Radar

In this case the signal is transmitted from Earth (station 1), received by the spacecraft (station 2), and then retransmitted by the spacecraft back to a different Earth-based station (station 3). This three-way communication allows for more precise tracking of the spacecraft by measuring the Doppler shift from both Earth stations. 

---

### Advantages and Disadvantages
**Advantages**:
- All-Weather, 24/7 Operation: radar signals are not significantly affected by clouds, weather, or sunlight. This allows for continuous tracking and observation, which is a major advantage over optical sensors
- Direct Range Measurement: unlike optical sensors, radar provides a direct and highly accurate measurement of the target's distance. This is crucial for initial orbit determination and for quickly acquiring a robust orbital solution
- Direct Velocity Measurement: the Doppler measurement provides direct information about the target's speed towards or away from the radar, which is a key component for refining the orbital state
- Effective for LEO: radar is most effective for tracking objects in LEO, where atmospheric drag is a significant perturbing force. Radar's ability to get frequent, accurate measurements makes it ideal for updating orbits in this dynamic environment
- Object Characterization: high-resolution imaging radars can not only track but also provide detailed images of the target, revealing its size, shape, and rotational state

**Disadvantages**:
- Limited Range for GEO: the signal-to-noise ratio of radar decreases with the square of the range, making it difficult and expensive to track objects in higher orbits like MEO and GEO. This is where optical sensors have an advantage
- High Cost and Complexity: radar systems, especially large, ground-based phased-array radars, are extremely expensive to build, operate, and maintain
- Less Angular Precision than Optical: while radar can measure angles, the angular precision of large, dedicated optical telescopes is generally superior
- Vulnerability to Jamming: as an active system, a radar's signals can be subject to electronic jamming or interference

### Range estimate and Light Time Delay

The range between the satellite and the tracking station is ideally expressed as

$$ \rho(t_r) = \sqrt{(\mathbf r_s(t_r)-\mathbf r_g(t_r)) \cdot  (\mathbf r_s(t_r)-\mathbf r_g(t_r))}$$

where:

- $\mathbf r_s(t)$ is spacecraft position
- $\mathbf r_g(t)$ is tracking-station position
- $t_r$ is reception time


When acquiring range measurement, we need to consider that the signal does not travel at infinite speed. Suppose a tracking station receives a signal at time $t_r$. The signal was emitted by the spacecraft at an earlier time ( $t_t $, time of transmission):

$$t_t = t_r-\tau$$

where $\tau$ is the signal **light time delay**, i.e. the time needed to the signal to reach the tracking station. During this interval the spacecraft moves: therefore, the spacecraft position used in the range calculation must be evaluated at the transmission time, not at the reception time.

The light-time-corrected one-way range is instead

$$ \rho(t_r) = \sqrt{(\mathbf r_s(t_r-\tau)-\mathbf r_g(t_r)) \cdot  (\mathbf r_s(t_r-\tau)-\mathbf r_g(t_r))} $$

with

$$\tau = \frac{\rho}{c}$$

and $c$ the speed of light. In this case, $t_r-\tau$ is the transmitting time, which happens before the receiving time.

Substituting the second equation into the first gives the implicit range equation:

$$ \tau = \frac{1}{c}\sqrt{(\mathbf r_s(t_r-\tau)-\mathbf r_g(t_r)) \cdot  (\mathbf r_s(t_r-\tau)-\mathbf r_g(t_r))}$$

This equation is implicit because the unknown $\tau$ appears both outside and inside the spacecraft-position evaluation. To solve this problem, we use iteration ignoring light time for the first estimate:

$$ \tau^0 = \frac{1}{c}\sqrt{(\mathbf r_s(t_r)-\mathbf r_g(t_r)) \cdot  (\mathbf r_s(t_r)-\mathbf r_g(t_r))} $$

Replacing this value into the implicit equation gives us a second estimate of $\tau$, and the process can be repeated until the difference of the estimates falls below a specific threshold. Because spacecraft velocities are much smaller than $c$, this iteration normally converges very rapidly.




For example, consider a range measurement between a tracking station and a LEO satellite taken at $t_r=1800\ \text{s}$. The iteration gives approximately:

$$\tau = 0.0402910807823\ \mapsto t_t = 1800-\tau = 1799.9597089192\ \text{s}$$

While the instantaneous range at receiving time is $\rho_i = 12078.874939\ km$, the corrected range is $\rho = 12078.962143\ km$; the range is corrected by an amount of $82.2\ m$.

The iteration converges in three steps; the sign and size of this correction depend on the projection of the spacecraft velocity along the line of sight.
