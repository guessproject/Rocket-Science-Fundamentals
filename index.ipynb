# Rocket Science Fundamentals

A hands-on course on the mathematics and physics behind spaceflight, from numerical methods to the determination of a satellite's orbit from real tracking data. Each lesson combines theory with interactive Python examples: change the inputs, re-run the cell and see how the results change.

The sections build on each other and are listed in the suggested order. *Dynamic Systems* and *Observation and Estimation* provide the mathematical tools, *Astrodynamics* describes how spacecraft move, and *Orbit Determination* brings everything together to estimate where a spacecraft actually is.

## Dynamic Systems
How to describe and simulate systems that evolve over time. The section starts with the numerical methods used to find roots and integrate differential equations (Euler, Runge-Kutta, adaptive step size), then introduces state-space models, the state transition matrix and the stability of equilibrium points, and ends by applying these tools to the non-linear two-body problem.

- [Numerical Methods](Dynamic_Systems/01-Numerical_Methods.ipynb)
- [Dynamic Systems and State Transition Matrix](Dynamic_Systems/02-Dynamic_Systems_and_State_Transition_Matrix.ipynb)
- [Stability of Dynamic Systems](Dynamic_Systems/03-Stability_of_Dynamic_Systems.ipynb)
- [Non-linear Systems: the Two-body Problem](Dynamic_Systems/04-Non_linear_Systems_the_Two_body_Problem.ipynb)

## Observation and Estimation
How to extract the best possible knowledge of a system from noisy and incomplete measurements. Starting from statistics and stochastic processes, the section develops least squares for linear and non-linear problems and then the Kalman filter, from a one-dimensional example to the extended Kalman filter used to track a moving target.

## Astrodynamics
The motion of natural and artificial bodies in space. The section covers the two-body problem and the ways to represent an orbit, the perturbations caused by the Earth's non-spherical shape, orbital maneuvers and transfers, interplanetary mission design with Lambert's problem and porkchop plots, lunar orbits, rendezvous and proximity operations, and the circular restricted three-body problem.

## Orbit Determination
Estimating a spacecraft's orbit from tracking data, combining the dynamics of *Astrodynamics* with the estimation techniques of *Observation and Estimation*. The section describes the force and measurement models, how process noise accounts for modelling errors, and the main tracking systems: optical telescopes, radars and GPS receivers. Practical examples use Ansys ODTK.
